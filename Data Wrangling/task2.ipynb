Next is the Job Title Column

# Job Title

First we skim the column with for loop

```python
#Skimming the column
for x,y in dataset2['Job Title'].value_counts().to_dict().items():
    print(x,y)
```

The most obvious mistake is that the column names didn't match the global schema. The data are also not yet normalized therefore we normalize the data according to what has been done in task 1 and change the column name

```python
#Change the column name from job title to Title
dataset2 = dataset2.rename(columns={'Job Title': 'Title'})

#Normalize the data like in dataset1

#Creating a function to normalize the title column
def normalizetitle(title):
    # normalize to upper case letters
    title = title.upper()
    # change AND to &
    title = title.replace(" AND ", " & ")
    # replace multiple spaces with a single space, also trim spaces on both side
    title = re.sub( '\s+', ' ', title).strip()
    return title

#Applying the function to the column
dataset2['Title'] = dataset2.Title.apply(lambda x: normalizetitle(x))
```

We're done with Title column, next Organisation column

# Organisation

First we skim the column with for loop

```python
#Skimming the column
for x,y in dataset2['Organisation'].value_counts().to_dict().items():
    print(x,y)
```

Again the most obvious problem is the column names didn't match the global schema. There's also some inconsistency and upper and lower cases problem. Therefore we change the column name and normalize the data like in task 1

```python
#Change the column name from job title to Title
dataset2 = dataset2.rename(columns={'Organisation': 'Company'})

#Normalize the data like in dataset1

# the following function is the clean and normalize the company field.
def normalizecompany(company):
    if pd.isnull(company):
        return company
    else:
        # normalize to upper case letters
        company = company.upper()
        # replace multiple spaces with a single space, also trim spaces on both side
        company = re.sub( '\s+', ' ', company).strip()
        # remove special character at the end
        company = re.sub(r"\W+$", "", company)
        # replace LIMITED with LTD
        company = company.replace("LIMITED", "LTD")
        # replace " AND " with " & " for processing 
        company = company.replace(" & ", " AND ")
        # remove all special characters except space and dot
        company = re.sub(r'([^\w\s\.]|_)', '', company)
        # change " AND " back to " & " 
        company = company.replace(" AND ", " & ")
        return company

#Applying the function to the column
dataset2['Company'] = dataset2.Company.apply(lambda x: normalizecompany(x))
```

Organisation column is done, next is location

First, the column name already match the global schema. However, the data didn't match what has been cleaned in dataset 1 therefore it is normalized and matched to the dataset 1

```python
#Defining function to normalize the category by uppercasing the category and deleting spaces + trailing spaces

def normalizecategory(category):
    #function to skip null is not needed because we known there's no null in this column    
    # normalize to upper case letters        
    category = category.upper()
    # replace " AND " with " & "
    category = category.replace(" AND ", " & ")
    return category

#Applying the function
dataset2['Category'] = dataset2.Category.apply(lambda x: normalizecategory(x))

#Code to correct the incorrect category names by replacing it with the correct names

dataset2.Category.replace({'FINANCE & ACCOUNTING': 'ACCOUNTING & FINANCE JOBS','INFORMATION TECHNOLOGY': 'IT JOBS',
                          'PR, ADVERTISING & MARKETING': 'PR, ADVERTISING & MARKETING JOBS'}, inplace=True)
```

# Salary Per Month

First we skim the column with for loop

Few problems, from the global schema opening date and opening time are combined into one column of OpenDate. The format for it is dateThour and it's in string format. Therefore, both column are combined into one OpenDate column with T in the middle. Then the former 2 columns are dropped

# DUPLICATES

There's a possibility that there are data conflicts in form of duplicates therefore we should check the combined dataframe again for duplicates. Like in task 1, we use 3 columns to search for duplicates which are Title, Location, and Category.

- Title is choosen because it's more likely that the name could be the same
- OpenDate and CloseDate is choosen because if the duplicated rows has the same name and has the same OpenDate and CloseDate with each other,
it's more likely that it's a duplication

```python
#Checking the merged dataset for duplicates based by Title, OpenDate, and CloseDate
merge[merge.duplicated(['Title', 'OpenDate', 'CloseDate'], keep=False)].sort_values(['Title'])
```

As we can see from the ouptut, there are a lot of duplicates in the merged dataset. I'm assuming that the data from the first dataset is the actual data and the data from the second dataset is the duplicates therefore, the second data is dropped.

```python
#Code to drop the duplicates while keeping the first one
merge = merge.drop_duplicates(subset=['Title', 'OpenDate', 'CloseDate'], keep='first')
```
