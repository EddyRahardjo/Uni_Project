# Question 2 Probability

Implement a Python function that simulates the experiment of choosing a fruit from 3 boxes

```python
#Importing the library for this question
import numpy as np

#Initializing a function to randomly select the box and the fruit
def fruit_random(n):
    #Initializing empty list
    box_result = []
    fruit_result = []
    #Initializing the content of the box
    box_true = [['apples','apples','apples','orange','orange','orange','orange','orange'],
                ['apples','apples','apples','apples','orange','orange','orange','orange'],['apples','orange']]
    #For loop to make the test function
    for i in range(n):
        #Selecting the box
        select_box = np.random.choice(3)
        #Selecting the fruit
        select_fruit = np.random.choice(box_true[select_box])
        #collecting the answer and append to the list
        if select_box == 0:
            box_result.append('red')
        elif select_box ==1:
            box_result.append('blue')
        elif select_box ==2:
            box_result.append('yellow')
        fruit_result.append(select_fruit)
    
    #Return the result
    return np.array(box_result,dtype=object), np.array(fruit_result,dtype=object)

```

```python
#Testing the function
fruit_random(4)
```

2. Formal Derivation 

If the picked fruit is an apple, what is the probability that it was picked from the yellow box?



$$P({Yellowbox}|{Apple}) = \text{Is the probability that the box was yellow while the fruit picked is an apple}$$

$$\text{Opening this equation, according to the bayes rule we get}$$

$$P({Yellowbox}|{Apple}) = \frac{P({Apple}|{Yellowbox})*P(Yellow)}{P(Apple)}$$

$$\text{The denominator in this equation, P(Apple) is broken down as}$$

$$\text{P(Apple) = P(Apple|Yellow)*P(Yellow) + P(Apple|Red)*P(Red) + P(Apple|Blue)*P(Blue)}$$

$$\text{The probability of P(Apple) came from the sum of all conditional probability that }$$ 

$$\text{the Apple came from each box multiplied by the probability that we choose that box. From the question, we know that}$$

$$\text{ In the red box we have 3 apples and 5 oranges, In the blue box we have 4 apples and 4 oranges, and in the yellow box we have 1 apples and 1 orange. Therefore,}$$

$$P(Apple|Yellow) = \frac{1}{2}, P(Apple|Red) = \frac{3}{8}, P(Apple|Blue) = \frac{4}{8}, P(Yellow) = P(Red) = P(Blue) = \frac{1}{3}$$

$$\text{With all the information we have, insert all of them to the prior equation we got}$$

$$P({Yellowbox}|{Apple}) = \frac{\frac{1}{2}\times\frac{1}{3}}{(\frac{1}{2}\times\frac{1}{3}) + (\frac{3}{8}\times\frac{1}{3}) + (\frac{1}{2}\times\frac{1}{3})} = \frac{4}{11}$$

$$\text{Therefore, we got that the probability the box was yellow if the fruit was an apple is } \frac{4}{11}$$


### Question 5 Expected Values

 Consider the following simple one-player game: the player first rolls a fair six-sided die and then she determines her score as the sum of the outcomes of a number of additional die roles, where the number of additionally rolled dice is equal to the number rolled with the first die.

```python
#Making the Function for the dice experiment
def dice_experiment(x):
    #Making the RNG variable for the random seed
    RNG = np.random.default_rng(seed=42)
    #Making the array to store the result
    final_result = np.zeros(x, dtype = int)
    #For loop to make the dice result
    for i in range(x):
        #Store the first dice result
        first_result = RNG.integers(1, 7)
        #Making the array to store the second dice result
        second_result = np.zeros(6)
        #For loop to roll the second dice and storing the result
        for m in range(first_result):
            dice_result = RNG.integers(1, 7)
            second_result[m] = dice_result
        #Counting the final result of the dice experiment
        final_result[i] = sum(second_result)
    #return the dice result
    return final_result

   
```

```python
# Result of dice experiment for 10,000 repetition
result = dice_experiment(100000)
result 
```

```python
# Calculating the mean of 10,000 repetition
mean = np.mean(result)
#Calculating the error of 10,000 repetition
moe = 1.96 * (np.std(result) / np.sqrt(100000))

print('E[Z] of the 10,000 repetition of the dice function', round(mean,5))
print('Error margin of the 10,000 repetition of the dice function', round(moe,5))
```
