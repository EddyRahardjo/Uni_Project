## Covariances

**4**  Assume we have four financial assets and their values (per dollar) in 20 years from today are the random variables X1, X2, X3, and X4. Assume further that their mean values and standard deviations are 100, 100, 50, 50 and 20, 22, 5, 5, respectively. Finally, assume that the correlation matrix between the variables is

$$
R = 
\begin{pmatrix}
1.0 & 0.9 & 0.0 & 0.0 \\
0.9 & 1.0 & -0.9 & -0.9 \\
0.0 & -0.9 & 1.0 & 0.5 \\
0.0 & -0.9 & 0.5 & 1.0 \\
\end{pmatrix}
$$

where the entry $R_{i,j}$ corresponds to $\rho(X_{i},X_{j}) = \text{Cov}(X_{i},X_{j})/(\text{Std}(X_{i})\text{Std}(X_j))$. Now we consider the following four investment portfolios, each investing 100 dollars today:

1. Investing 50 into asset 1, 0 into asset 2, 25 into asset 3, and 25 into assset 4

2. 0, 50, 50, 0

3. 0, 50, 25, 25

4. 25, 25, 25, 25

4.1 Compute the expected values and standard deviations of each portfolio after 20 years using Python (with only imports from numpy) and give a recommendation in a markdown cell on what portfolio a conservative investor should choose.


```python

#Import the library
import numpy as np

#Intializing R, mean, and std
R = np.array([[1.0, 0.9, 0.0, 0.0], [0.9, 1.0, -0.9, -0.9], [0.0, -0.9, 1.0, 0.5], [0.0, -0.9, 0.5, 1.0]])

mu = np.array([100, 100, 50, 50])

std = np.array([20, 22, 5, 5])

#Counting covariance matrix
cov = []

for i in range(4):
    for j in range(4):
        cov.append(R[i, j] * std[i] * std[j])
#converting the covariance into numpy
cov = np.array(cov)
#Reshaping the array into 4,4 array
cov = (cov.reshape(4,4))

#Scenario 1 Expected Value
port1 = np.array([50, 0, 25, 25])
expected1 = np.dot(mu, port1)

#Scenario 2 Expected Value
port2 = np.array([0, 50, 50, 0])
expected2 = np.dot(mu, port2)

#Scenario 3 Expected Value
port3 = np.array([0, 50, 25, 25])
expected3 = np.dot(mu, port3)

#Scenario 4 Expected Value
port4 = np.array([25, 25, 25, 25])
expected4 = np.dot(mu, port4)

#Checking the mean for all scenario
print('Scenario 1 mean = ', expected1, 'Scenario 2 mean = ',expected2, 'Scenario 3 mean = ',expected3, 'Scenario 4 mean = ',expected4)

#Scenario 1 Weighted Covariance
w_cov1 = np.dot(port1.T, np.dot(cov, port1))

#Scenario 2 Weighted Covariance
w_cov2 = np.dot(port2.T, np.dot(cov, port2))

#Scenario 3 Weighted Covariance
w_cov3 = np.dot(port3.T, np.dot(cov, port3))

#Scenario 4 Weighted Covariance
w_cov4 = np.dot(port4.T, np.dot(cov, port4))

print('Scenario 1 Covariance = ', np.sqrt(w_cov1), 'Scenario 2 Covariance = ', np.sqrt(w_cov2), 
      'Scenario 3 Covariance = ', np.sqrt(w_cov3), 'Scenario 4 Covariance = ', np.sqrt(w_cov4))
```

Based on the mean and covariance, the best scenario is scenario 3.
